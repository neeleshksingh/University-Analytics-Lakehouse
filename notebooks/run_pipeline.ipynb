```python
import importlib
import os
import sys

dbutils.widgets.text("layer", "bronze")          # the Job passes this in
layer = dbutils.widgets.get("layer")

os.environ["LAKEHOUSE_TARGET"] = "databricks"
os.environ["UC_CATALOG"] = "workspace"
sys.path.insert(0, os.path.abspath(".."))

LAYERS = {"bronze", "silver", "gold"}
if layer not in LAYERS:
    raise ValueError(f"layer must be one of {sorted(LAYERS)}, got {layer!r}")

print(f"Running layer: {layer}")
importlib.import_module(f"src.lakehouse.{layer}").main()
```
